$CRC 32$

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = pd.read_csv('dispersia_1.csv') 

x = data['bucket_num']
y = data['bucket_size']

# -----------------------
dispersia = np.var(y)
# -----------------------

df = pd.DataFrame(data)

plt.figure(figsize=(10, 6))
plt.bar(x, y)

bars = plt.bar(
    df["bucket_num"],
    df["bucket_size"],
    width=1,          # Почти перекрывающиеся столбцы
    color='#1f77b4',    # Синий цвет
    label=f'дисперсия = {dispersia:.3f}'
)

plt.title("Распределение элементов по бакетам в хеш-таблице (CRC32)")
plt.xlabel("Номер бакета")
plt.ylabel("Кол-во элементов в бакете")

plt.ylim(0, 25)
plt.yticks(range(0, 25, 1)) 

plt.grid(
    True,
    color='lightblue',
    linestyle='--',
    alpha=0.6
)

plt.legend()
plt.tight_layout()
plt.show()
```
